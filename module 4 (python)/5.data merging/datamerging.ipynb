```python
# identifying missing values using isnull() or isna():

import pandas as pd

# creating a sample dataframe
data = {
    'Num':[1,2,None,4,5],
    'alph':['A','B',None,'D','E'],
    'alpha':['x','y','z',None,'w']
}
df= pd.DataFrame(data)
df

# checking for missing values:
missing_values= df.isnull()
print(missing_values)

# counting missing values in each column:
missing_values_count= df.isnull().sum()
print(missing_values_count)

# checking if any missing values exist in the dataframe:
check_missing_values= df.isnull().values.any()
check_missing_values
```

```python
# removing rows or columns with missing values using dropna():

# dropping rows with any missing values:
df_without_missing= df.dropna()
print(df_without_missing)

# dropping columns with any missing values:
df_without_missingc=df.dropna(axis=1)
print(df_without_missingc)

# dropping rows with missing values in specfic columns:
df_without_missingsc= df.dropna(subset=['Num','alph'])
df_without_missingsc
```

```python
# filling missing values using fillna() with methods like mean, median, and forward/backward filling:

# filling missing values with mean of each column:
df1= df.fillna(df.mean)
df1
# filling missing values with median of each column:
df1= df.fillna(df.median)
df1
# filling missing values using forward filling(propagating last valid observation forward)
df3= df.ffill()
df3
# filling missing values using backward filling(propagating next valid observation forward)
df3= df.bfill()
df3
```

```python
# standardizing data  formats using pandas' string manipulation functions (str.lower(),str.upper(),str.replace()):

# creating a sample dataframe
data= {'column_name':['value1','value2','old_string','value4']}
df1= pd.DataFrame(data)
df1

# converting column values to lowercase:
df1['column_name']= df1['column_name'].str.lower()
df1['column_name']

# converting column values to uppercase:
# df1['column_name']= df1['column_name'].str.upper()
# df1['column_name']

# replacing a specific substring in column values:
df1['column_name']= df1['column_name'].str.replace('old_string','new_string')
df1['column_name']
```

```python
# correcting inconsistent data using replace() function:

df=df1.copy()
# assuring 'df' is your dataframe and 'column_name' is the name of the column you want to correct:
# replacing  specific values with new values:
df['column_name']= df['column_name'].replace({'value1':'A','value2':'B'})
df['column_name']
# replacing values based on a condition:
df['column_name']=df['column_name'].replace('new_string','C', regex=True)
df
# replacing values using regular expressions (re module): 
import re
df['column_name']=df['column_name'].apply(lambda x: re.sub(r'pattern', 'replacement', x))
df['column_name']
```

```python
# detecting outliers using z-score and iqr

import pandas as pd
import numpy as np

# creating a sample dataframe:
data={'nu':[10,15,12,8,20,25,7,300000]}
df=pd.DataFrame(data)
df
# calculating z-score for the column:
z_scores = (df['nu']-df['nu'].mean())/df['nu'].std()

# defining a threshold for outlier detection(e.g, z-score>3 or < -3):
threshold = 3
# identifying outliers based on the threshold:
outliers= df[abs(z_scores)>threshold]
# checking if outliers dataframe is empty:
if outliers.empty:
    print('No outliers')
else:
    print('outliers')
```

```python
import pandas as pd
import numpy as np

# creating a sample dataframe:
data={'na':[10,15,12,8,20,25,7,300000]}
df=pd.DataFrame(data)
df
# calculating the first quartile(Q1):
q1= df['na'].quantile(0.25)
q1
# calculating the third  quartile(Q3):
q3= df['na'].quantile(0.75)
q3
# calculating the IQR:
IQR= q3-q1
IQR
# defining a threshold for outlier detection(e.g, values outside the range[ q1-1.5*IQR, q3+1.5*IQR] are considered):
lower_threshold= q1-1.5*IQR
lower_threshold
upper_threshold= q3+1.5*IQR
upper_threshold
# identifying outliers based on the threshold:
outliers= df[(df['na']<lower_threshold) | (df['na']>upper_threshold)]
outliers
# checking if outliers dataframe is empty:
if outliers.empty:
    print('No outliers')
else:
    print('outliers')
```

```python
import pandas as pd

# creating sample dataframes
df1=pd.DataFrame({'id':[1,2,3,4],
                  'name':['john','alice','bob','emily']})
df2=pd.DataFrame({'id':[2,3,5,6],
                  'city':['new york','london','paris','berlin']})

# inner join: returns the intersection of the two datasets based on a common key:
inner_join= pd.merge(df1,df2,on='id',how='inner')
inner_join

# left join: returns all records from the left dataset and matching records from the right dataset:
left_join= pd.merge(df1,df2,on='id',how='left')
left_join

# right join: returns all records from the right dataset and matching records from the left dataset:
right_join= pd.merge(df1,df2,on='id',how='right')
right_join

# outer join: returns all records when there is a match in either the left or right dataset:
full_join= pd.merge(df1,df2,on='id',how='outer')
full_join

```
