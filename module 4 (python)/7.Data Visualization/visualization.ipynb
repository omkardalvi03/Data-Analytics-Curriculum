Different kinds of plots

1. LINE PLOT

```python
#line plot
import pandas as pd
import matplotlib.pyplot as plt

# create a dataframe
data= {'year':[2010,2011,2012,2013,2014,2015],
      'sales':[100,150,120,180,200,250]}
df=pd.DataFrame(data)
df

# create a line plot
plt.plot(df['year'],df['sales'])
plt.xlabel('year')
plt.ylabel('sales')
plt.title('sales over the year')

# display the plot
plt.show()
```

```python
# multiple lines
data= {'year':[2010,2011,2012,2013,2014,2015],
      'sales':[100,150,120,180,200,250],
      'sales2':[80,120,110,160,190,220]}
df=pd.DataFrame(data)
df

# create a line plot
plt.plot(df['year'],df['sales'],label='sales')
plt.plot(df['year'],df['sales2'],label='sales2')
plt.xlabel('year')
plt.ylabel('sales')
plt.title('sales over the year')
plt.legend()

# display the plot
plt.show()
```

2. BAR PLOT

```python
#bar plot
# create a dataframe
data= {'year':[2010,2011,2012,2013,2014,2015],
      'sales':[100,150,120,180,200,250]}
df=pd.DataFrame(data)
df

# CREATE A BAR PLOT
plt.bar(df['year'],df['sales'])
plt.xlabel('year')
plt.ylabel('sales')
plt.title('sales by year')

# display the plot
plt.show()
```

3. SCATTER PLOTS

```python
# scatter plot
data={'x':[1,2,3,4,5],
      'y':[2,4,6,8,10]}
df=pd.DataFrame(data)
df

# create a scatterplot
plt.scatter(df['x'],df['y'])
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter Plot')

# display the plot
plt.show()
```

```python
# with a trend line
import seaborn as sns

# create a scatterplot with a trend line
sns.regplot(x='x', y='y', data=df)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter Plot with Trend line')

# display the plot
plt.show()
```

4. HISTOGRAM

```python
import numpy as np

# generate random data
np.random.seed(42)
data= np.random.normal(loc=50, scale=10, size=1000)

# create a dataframe
df= pd.DataFrame({'values': data})

# create a histogram
plt.hist(df['values'], bins=20, color='skyblue', edgecolor= 'black')
plt.xlabel('values')
plt.ylabel('frequency')
plt.title('Histogram')

# display the plot
plt.show()
```

5. BOX PLOTS

```python
data= {'category':['A','A','A','B','B','B','B'],
       'value':[10,15,20,25,30,35,40]}
df= pd.DataFrame(data)
df


sns.boxplot(x='category',y='value', data=df)
plt.xlabel('category')
plt.ylabel('value')
plt.title('BOX PLOT')

# display the plot
plt.show()
```

6. HEATMAPS

```python
data={'A':[1,2,3],
      'B':[4,5,6],
      'C':[7,8,9]}
df=pd.DataFrame(data)
df

sns.heatmap(df, annot=True, cmap='YlGnBu')
plt.xlabel('Columns')
plt.ylabel('Rows')
plt.title('HEATMAP')

# display the plot
plt.show()
```
