```python
import pandas as pd

# creating a dataframe

data={
    'Name':['john','jane','steve','linda','jane'],
    'Age':[25,35,None,28,35],
    'Salary':[50000,60000,45000,None,60000],
    'Department':['marketing','sales','marketing','engi','sales']
}
df = pd.DataFrame(data)
df

# handle missing data
df_cleaned= df.dropna()
df_cleaned

df_filled= df.fillna(0)
df_filled

# data filtering
filtered_df= df[df.Salary>0]
filtered_df

# data sorting
sorted_df= df.sort_values('Age', ascending=False)
sorted_df

# data grouping and aggregation 
grouped_df = df.groupby('Age').mean(numeric_only=True)
grouped_df

groupe_df = df.groupby('Department').mean(numeric_only=True)
groupe_df

# drop duplicates

df_cleaned= df.drop_duplicates()
df_cleaned

# data I/O 
df.to_csv('data.csv',index=False)
df

df.to_excel('data.xlsx',index=False)
df_excel= pd.read_excel('data.xlsx')
df_excel
# df_sql= pd.read_sql_query('SELECT * FROM TABLE', 'sqlite://data.db')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

data={
    'Name':['john','jane','steve','linda'],
    'Age':[25,35,None,28],
    'Salary':[50000,60000,45000,None],
    'Department':['marketing','sales','marketing','engi'],
    'Gender':['M','F','M','F'],
    'City':['NYC','Chicago','LA','Chicago']
}
df=pd.DataFrame(data)
df

# descriptive statistics
summary_stats= df.describe()
summary_stats

# data visualization
sns.set(style='darkgrid')

# bar plot
sns.barplot(x='Department', y='Salary', data=df)
plt.title('Average salary by Department')
plt.show()

# data preprocessing 
df['Department']= df['Department'].str.upper()
print(df)

# data transformation
df['Salary in months']= df['Salary']/12
print(df)

# data visualization using matplotlib

plt.scatter(df['Age'],df['Salary'])
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
plt.show()
```
