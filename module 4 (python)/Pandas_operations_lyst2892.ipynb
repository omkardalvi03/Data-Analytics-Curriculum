```python
import pandas as pd

# Creating a dataframe

data={
    'Name':['John','Jane','Steve','Linda','Jane'],
    'Age':[25,35,None,28,35],
    'Salary':[50000,60000,45000,None,60000],
    'Dept':['Mktng','Sales','Mktng','Engg','Sales']
}
df= pd.DataFrame(data)
df

# Handle missing data

df_cleaned= df.dropna()
df_filled= df.fillna(0)

# Data filtering

filtered_df= df[df.Salary>0]
filtered_df

# Data sorting

sorted_df= df.sort_values('Age', ascending=False)
sorted_df

# Data grouping and aggregation

grouped_df= df.groupby('Dept').mean(numeric_only=True)
grouped_df

# Drop duplicates

df_cleaned= df.drop_duplicates()
df_cleaned\

# Data I/O

df.to_excel('data.xlsx', index=False)
df_excel= pd.read_excel('data.xlsx')
# df_sql= pd.read_sql_query('SELECT * FROM table', 'sqlite://data.db')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

data={
    'Name':['John','Jane','Steve','Linda'],
    'Age':[25,35,None,28],
    'Salary':[50000,60000,45000,None],
    'Dept':['Mktng','Sales','Mktng','Engg'],
    'Gender':['M','F','M','F'],
    'City':['NYC','Chicago','LA','Chicago']
}
df= pd.DataFrame(data)
df


# Descriptive statistics

summary_stats= df.describe()
summary_stats

# Data visualization

sns.set(style='darkgrid')

# Bar plot

sns.barplot(x='Dept',y='Salary', data=df)
plt.title('Average salary by Department')
plt.show()

# Data preprocessing

df['Dept']= df['Dept'].str.upper()
df

# Data transformation

df['Salary in months']= df['Salary']/12
df

# Data Visualization using Matplotlib\

plt.scatter(df['Age'], df['Salary'], color='blue')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age VS Salary')
plt.show()
```
