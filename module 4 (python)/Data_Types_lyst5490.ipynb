```python
# Numeric Types and Arithmetic Operators

# Integer
num1 = 10

# Floating-point number
num2 = 3.14

# Complex number
num3 = 2 + 3j

# Addition
result_add = num1 + num2

# Subtraction
result_sub = num1 - num2

# Multiplication
result_mul = num1 * num2

# Division
result_div = num1 / num2

# Modulus (remainder)
result_mod = num1 % 3

# Exponentiation
result_exp = num1 ** 2

# Print the results
print("Addition:", result_add)
print("Subtraction:", result_sub)
print("Multiplication:", result_mul)
print("Division:", result_div)
print("Modulus:", result_mod)
print("Exponentiation:", result_exp)
```

```python
# Sequence Types and Indexing

# Create a list
fruits = ['apple', 'banana', 'cherry', 'date', 'elderberry']

# Accessing elements by index
first_fruit = fruits[0]
second_fruit = fruits[1]
last_fruit = fruits[-1]

# Slicing a list
sliced_fruits = fruits[1:4]

# Modifying list elements
fruits[2] = 'grape'

# Adding elements to a list
fruits.append('fig')
fruits.insert(2, 'orange')

# Removing elements from a list
fruits.remove('banana')
popped_fruit = fruits.pop(3)

# Print the list and results
print("Fruits:", fruits)
print("First fruit:", first_fruit)
print("Second fruit:", second_fruit)
print("Last fruit:", last_fruit)
print("Sliced fruits:", sliced_fruits)
print("Popped fruit:", popped_fruit)
```

```python
# String Manipulation and Comparison

# String concatenation
name = "John"
greeting = "Hello, " + name + "!"

# String length
name_length = len(name)

# String indexing
first_letter = name[0]
last_letter = name[-1]

# String slicing
sliced_name = name[1:3]

# String comparison
is_same = name == "John"

# String methods
uppercase_name = name.upper()
lowercase_name = name.lower()
capitalized_name = name.capitalize()

# Print the results
print("Greeting:", greeting)
print("Name length:", name_length)
print("First letter:", first_letter)
print("Last letter:", last_letter)
print("Sliced name:", sliced_name)
print("Is same:", is_same)
print("Uppercase name:", uppercase_name)
print("Lowercase name:", lowercase_name)
print("Capitalized name:", capitalized_name)
```

```python
# Boolean Logic and Logical Operators

# Boolean variables
is_sunny = True
is_raining = False

# Logical NOT
not_sunny = not is_sunny

# Logical AND
both_sunny_and_raining = is_sunny and is_raining

# Logical OR
either_sunny_or_raining = is_sunny or is_raining

# Print the results
print("Is sunny:", is_sunny)
print("Is raining:", is_raining)
print("NOT sunny:", not_sunny)
print("Both sunny and raining:", both_sunny_and_raining)
print("Either sunny or raining:", either_sunny_or_raining)
```
