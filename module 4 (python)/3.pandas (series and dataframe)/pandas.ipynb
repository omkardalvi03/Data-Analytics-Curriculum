```python
# series

import pandas as pd

data=[10,20,30,40,50]
index=['A','B','C','D','E']

series= pd.Series(data,index)
print(series)

element_a= series['A']
element_b= series['D']

print(element_a)
print(element_b)
```

```python
# slicing 
slice_1= series[1:4]
print(slice_1)

slice_2= series['B':'E']
print(slice_2)

# airthmetic operations
series1= pd.Series([1,2,3])
series2= pd.Series([4,5,6])
result= series1+series2
print(result)

# applying functions
def square(number):
    return number **2
squared_series= series.apply(square)
squared_series
```

```python
# dataframe

data={'Name':['john','alice','bob'],
      'Age':[25,30,35],
      'Country':['Usa','Canada','UK']}
df= pd.DataFrame(data)
print(df)

name= df['Name']
name
```

```python
# accessing rows using loc:
first_row= df.loc[0]
print(first_row)
third_row= df.loc[2]
print(third_row)

# accessing rows using iloc:
first_row= df.iloc[0]
print(first_row)
third_row= df.iloc[2]
print(third_row)

# adding a new column:
df['salary']=[50000,60000,70000]
print(df)


#filtering rows based on a condition:
filtered_df= df[df['Age']>=30]
print(filtered_df)

filtere_df= df[df['salary']<60000]
print(filtere_df)

#filtering functions to a dataframes:
def uppercase_names(row):
    return row['Name'].upper()

df['upper name']= df.apply(uppercase_names, axis=1)
df
```
